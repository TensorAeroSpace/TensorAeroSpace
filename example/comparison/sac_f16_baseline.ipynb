# SAC Baseline on F-16

This notebook trains and evaluates a SAC agent on the `LinearLongitudinalF16-v0` environment for angle-of-attack (alpha) tracking with a 5-degree step reference.

```python
import gymnasium as gym
import numpy as np
import torch

from tensoraerospace.agent.sac import SAC
from tensoraerospace.signals.standart import unit_step
from tensoraerospace.utils import convert_tp_to_sec_tp, generate_time_period
```

## 2. Simulation Parameters

Set up discretization, time period, and the step reference signal.

```python
dt = 0.01  # Discretization
tp = generate_time_period(tn=20, dt=dt) # Time period
tps = convert_tp_to_sec_tp(tp, dt=dt)
number_time_steps = len(tp) # Number of time steps
reference_signals = np.reshape(unit_step(degree=5, tp=tp, time_step=10*dt, output_rad=True), [1, -1]) # Reference signal
```

## 3. Environment Setup

Create the F-16 linear longitudinal environment tracking alpha.

```python
env = gym.make(
    "LinearLongitudinalF16-v0",
    number_time_steps=number_time_steps,
    use_reward=True,
    initial_state=[[0], [0], [0]],
    reference_signal=reference_signals,
    state_space=["theta", "alpha", "q"],
    output_space=["theta", "alpha", "q"],
    tracking_states=["alpha"],
)

env.reset()
```

```python
seed = 42
replay_size = 1000000
batch_size = 256
updates_per_step = 1
num_steps = 1000001
```

```python
torch.manual_seed(seed)
np.random.seed(seed)
```

## 4. SAC Agent Setup and Training

Configure and train the SAC agent.

```python
agent = SAC(env, memory_capacity=replay_size, hidden_size=32, device="cpu", verbose_histogram=False)
```

```python
agent.train(num_episodes=1)
```

## 5. Evaluation

Run evaluation episodes and collect rewards.

```python
# Evaluation
state, info = env.reset()
done = False
total_rew = 0
while not done:
    action = agent.select_action(state)
    state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
    total_rew+=reward
print(total_rew)
```

```python
num_eval_episodes = 3
episode_returns = []
episode_lengths = []

for episode_num in range(num_eval_episodes):
    state, info = env.reset()
    done = False
    total_rew = 0.0
    steps = 0

    while not done:
        action = agent.select_action(state)
        state, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated
        total_rew += float(reward)
        steps += 1

    episode_returns.append(total_rew)
    episode_lengths.append(steps)

print(f"Episode total rewards: {episode_returns}")
print(f"Episode lengths: {episode_lengths}")
```

## 6. Results

Plot the alpha transient response.

```python
env.unwrapped.model.plot_transient_process('alpha', tps, reference_signals[0], to_deg=True, figsize=(15,4))
```
