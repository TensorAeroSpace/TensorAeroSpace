# 🛰️ Example: Working with ImprovedComSatEnv

This notebook demonstrates the usage of the improved communication satellite environment with normalized spaces.

## 📋 Features of ImprovedComSatEnv:
- ✅ Normalized action and observation spaces [-1, 1]
- ✅ Improved reward function
- ✅ Optimized for reinforcement learning
- ✅ Thrust control for orbit correction


## 📚 Importing Libraries

```python
import gymnasium as gym
import numpy as np
import tensoraerospace
from tensoraerospace.utils import generate_time_period, convert_tp_to_sec_tp
from tensoraerospace.signals.standart import unit_step
```

## ⚙️ Parameters Setup

```python
dt = 0.01
tp = generate_time_period(tn=20, dt=dt)
number_time_steps = len(tp)
reference_signals = np.reshape(
    unit_step(degree=5, tp=tp, time_step=10, output_rad=True), 
    [1, -1]
)
print(f"📊 Параметры: время={tp[-1]:.1f}с, шагов={number_time_steps}")
```

## 🚀 Creating the Environment

```python
# Начальное состояние [rho, rho_dot, theta_dot] в км, м/с, рад/с
initial_state = np.array([6371.0, 0.0, 0.0])

env = gym.make(
    "ImprovedComSat-v0",
    initial_state=initial_state,
    reference_signal=reference_signals,
    number_time_steps=number_time_steps,
    dt=dt,
    initial_thrust=0.0,
    use_initial_action_on_first_step=True
)

obs, info = env.reset()
print(f"✅ Среда создана!")
print(f"📊 Наблюдение: {obs}")
print(f"🎯 Действия: {env.action_space}")
print(f"📐 Наблюдения: {env.observation_space}")
```

## 🎮 Executing a Step

```python
action = np.array([0.3], dtype=np.float32)
obs, reward, terminated, truncated, info = env.step(action)
print(f"🎯 Действие: {action[0]:.2f}")
print(f"📊 Наблюдение: {obs}")
print(f"🏆 Награда: {reward:.6f}")
```
