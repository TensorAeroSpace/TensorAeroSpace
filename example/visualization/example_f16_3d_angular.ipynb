# 3D Flight Visualization — Nonlinear F-16 (6-DoF)

Demonstrates the new `env.render()` API on the angular F-16 environment.
Renders an open-loop banked turn and visualises it with a 3D trail and
time-synced state charts. Drag the 3D scene with the mouse to rotate the camera.

```
import gymnasium as gym
import numpy as np
import tensoraerospace  # registers env_ids
```

```
env = gym.make(
    "NonlinearAngularF16-v0",
    initial_state=np.zeros(14),
    number_time_steps=300,
    dt=0.02,
    airspeed=220.0,
    render_mode="human",
)
obs, _ = env.reset()
# Open-loop control: small banked turn (constant aileron + elevator)
action = np.array([2.0, 5.0, 0.0])  # stab, ail, dir (degrees)
for _ in range(300):
    obs, _, _, truncated, _ = env.step(action)
    if truncated:
        break
```

```
fig = env.render()
fig.show()
```
