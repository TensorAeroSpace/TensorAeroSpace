# 3D Flight Visualization — Nonlinear F-16 (Longitudinal, 2-DoF)

Demonstrates the new `env.render()` API on the longitudinal F-16
environment. The trail collapses to the vertical (x-z) plane since
lateral motion is zero. The pitch-up command in the first half of the
rollout creates a visible climb, then the elevator returns to neutral.

```
import gymnasium as gym
import numpy as np
import tensoraerospace  # registers env_ids
```

```
ref = np.zeros((1, 200))
env = gym.make(
    "NonlinearLongitudinalF16-v0",
    initial_state=np.zeros(4),
    reference_signal=ref,
    number_time_steps=200,
    dt=0.02,
    airspeed=200.0,
    render_mode="human",
)
env.reset()
# Open-loop: pitch-up for the first half, neutral for the rest.
for k in range(200):
    action = np.array([4.0 if k < 100 else 0.0])
    env.step(action)
```

```
fig = env.render()
fig.show()
```
